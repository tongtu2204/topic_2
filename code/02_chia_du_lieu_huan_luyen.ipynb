```python
import pandas as pd
from pathlib import Path

PROCESSED_DIR = Path("D:/Chuyên đề 2/data/processed")
```

```python
# ============================================================
# 1. ĐỌC DATA GIÁ SẠCH
# ============================================================

price = pd.read_csv(
    PROCESSED_DIR / "silver_price_clean.csv",
    parse_dates=["Date"]
)

# ============================================================
# 2. ĐỌC DATA RETURN SẠCH
# ============================================================

returns = pd.read_csv(
    PROCESSED_DIR / "silver_return_clean.csv",
    parse_dates=["Date"]
)
```

```python
# ============================================================
# 3. CHIA TRAIN / VALIDATION / TEST
# ============================================================

train_end = "2024-12-31"
val_end = "2025-12-31"

# PRICE
price_train = price[
    price["Date"] <= train_end
].copy()

price_val = price[
    (price["Date"] > train_end) &
    (price["Date"] <= val_end)
].copy()

price_test = price[
    price["Date"] > val_end
].copy()

# RETURN
return_train = returns[
    returns["Date"] <= train_end
].copy()

return_val = returns[
    (returns["Date"] > train_end) &
    (returns["Date"] <= val_end)
].copy()

return_test = returns[
    returns["Date"] > val_end
].copy()
```

```python
# ============================================================
# 4. IN KÍCH THƯỚC
# ============================================================

print("PRICE")
print("Train:", price_train.shape)
print("Validation:", price_val.shape)
print("Test:", price_test.shape)

print("\nRETURN")
print("Train:", return_train.shape)
print("Validation:", return_val.shape)
print("Test:", return_test.shape)
```

```python
# ============================================================
# 5. LƯU
# ============================================================

price_train.to_csv(
    PROCESSED_DIR / "price_train.csv",
    index=False
)

price_val.to_csv(
    PROCESSED_DIR / "price_validation.csv",
    index=False
)

price_test.to_csv(
    PROCESSED_DIR / "price_test.csv",
    index=False
)

return_train.to_csv(
    PROCESSED_DIR / "return_train.csv",
    index=False
)

return_val.to_csv(
    PROCESSED_DIR / "return_validation.csv",
    index=False
)

return_test.to_csv(
    PROCESSED_DIR / "return_test.csv",
    index=False
)

print("\nĐã lưu toàn bộ Train / Validation / Test.")
```
